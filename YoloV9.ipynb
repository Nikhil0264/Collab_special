```python
from ultralytics import YOLO
model = YOLO('yolov9e.pt')
```

```python
model.info(verbose=True)
```

```python
print(model.model)
```

```python
image_path='/pngtree-fresh-fruit-juices-drinks-3d-png-image_12270313.png'
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(image_path)
result = model(image)
```

```python
for r in result:
  r.save(filename='result.jpg')
```

```python
imag_result = cv2.imread('result.jpg')
img_rgb = cv2.cvtColor(imag_result, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10,10))
plt.imshow(img_rgb)
plt.show()
```
