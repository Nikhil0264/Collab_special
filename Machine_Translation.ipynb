```python
class ruleBase_Translator:
  def __init__(self):
    self.translation_rules = {
        'lexical_map':{
            'en':{
                'hello':'bonjour',
                'bye':'au revoir',
                'thank you':'merci',
                'please':'s',
                'yes':'oui',
                'no':'non',
                'world':'monde'
            },
            'grammer_rules':{
                'adj_noun_swap': lambda adj,noun:f"{noun} {adj}"
            }
        }
    }

  def translate(self,text,source_lan='en',target_lan='fr'):
    tokens = text.lower().split()

    translated = {
        self.translation_rules['lexical_map'][source_lan].get(token,token)
        for token in tokens
    }
    return ' '.join(translated)
```

```python
trans = ruleBase_Translator()
trans.translate('hello world')
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def simple_sequence_prediction():

  x = np.array([[i+j for j in range(3)] for i in range(100)])
  y = np.array([i+3 for i in range(100)])

  X = x.reshape((100,3,1))

  model = Sequential(
      [
          SimpleRNN(units=32,input_shape=(3,1)),
          Dense(units=1)
      ]
  )

  model.compile(optimizer='adam',loss='mse')
  model.fit(X,y,epochs=100,verbose=0)

  test_sequence = np.array([10,11,12]).reshape(1,3,1)
  prediction = model.predict(test_sequence)
  y_hat = prediction[0][0]
  print(f"Sequence {test_sequence.flatten()} ---> Predicted next number: {np.round(y_hat)}")
```

```python
simple_sequence_prediction()
```
