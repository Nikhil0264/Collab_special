````python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1-0528",
    max_new_tokens=512,
    do_sample=False,
    repetition_penalty=1.03,
    provider="auto",  # let Hugging Face choose the best provider for you
)

chat_model = ChatHuggingFace(llm=llm)
````

````python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in 30 words."
)
````

````python
prompt = ChatPromptTemplate.from_template(
    "I want to open a restaurant for {cuisine} food.Suggest some names for this ?"
)
chain = prompt | chat_model
name_chain = chain.invoke ({"cuisine" : "Indian"})
prompt = ChatPromptTemplate.from_template(
    "Suggest some new items for {restaurant_name}"
)
main_chain = prompt | chat_model
response = main_chain.invoke ({"restaurant_name" : name_chain.content})
Markdown(response.content)
````

````python
from langchain_community.utilities import SerpAPIWrapper
from langchain_core.tools import tool

search = SerpAPIWrapper(serpapi_api_key=Search_Key)

@tool
def serpapi_search(query: str) -> str:
    """Search the web using SerpAPI and return search results."""
    return search.run(query)
````

````python
from langchain.agents import create_agent

agent = create_agent(
    model=chat_model,
    tools=[serpapi_search],
    system_prompt="""
    You are a helpful research assistant.
    Use the SerpAPI search tool whenever current or
    up-to-date information is required.
    """
)
````

````python
response = agent.invoke(
    {
        "messages": [
            ("user", "What is the GDP of the US in 2026?")
        ]
    },
    {
        "verbose": True
    }
)

print(response["messages"][-1].content)
````
